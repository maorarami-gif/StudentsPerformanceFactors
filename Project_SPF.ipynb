```python
df = spf_df[["Parental_Involvement","Parental_Education_Level","Family_Income","Exam_Score"]] 

g = sns.FacetGrid(data=df,row="Parental_Education_Level",height=2.2,aspect=2.2)

g.map_dataframe(sns.boxplot,x="Parental_Involvement",y="Exam_Score",hue="Family_Income",palette="Set2")

g.set_axis_labels("Parental Involvement", "Exam Score")

g.set_titles(row_template="Parental Education Level: {row_name}")

for ax in g.axes.flatten():  
    ax.legend(loc="upper left")
plt.tight_layout()
plt.show()
```

```python
df = spf_df[["Parental_Involvement","Parental_Education_Level","Exam_Score"]]

pivot = df.pivot_table(index="Parental_Education_Level",columns="Parental_Involvement",values="Exam_Score",aggfunc="mean")

plt.figure(figsize=(6,4))
sns.heatmap(pivot, annot=True, fmt=".1f", cmap="Blues")
plt.title("Mean Exam Score by Parental Involvement and Education Level")
plt.xlabel("Parental Involvement")
plt.ylabel("Parental Education Level")
plt.tight_layout()
plt.show()
```
